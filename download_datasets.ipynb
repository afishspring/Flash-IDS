```python
import os
import gdown
import subprocess
import zipfile
import tarfile

def download_from_drive(url, output_path):
    """
    使用 gdown 从 Google Drive 下载文件。
    如果 output_path 已经存在，则跳过下载。
    """
    if os.path.exists(output_path):
        print(f"[INFO] 文件 {output_path} 已存在，跳过下载。")
        return
    
    # 确保父目录存在
    directory = os.path.dirname(output_path)
    if not os.path.exists(directory):
        os.makedirs(directory)
        
    print(f"[INFO] 正在下载 {output_path} ...")
    # quiet=False 显示进度条，fuzzy=True 允许更宽松的 URL 匹配
    gdown.download(url, output_path, quiet=False, fuzzy=True)

def extract_tar_gz(file_path, extract_path):
    """
    解压 .tar.gz 文件到指定目录。
    """
    if not os.path.exists(file_path):
        return
    
    print(f"[INFO] 正在解压 {file_path} 到 {extract_path} ...")
    try:
        with tarfile.open(file_path, "r:gz") as tar:
            tar.extractall(path=extract_path)
        print("[INFO] 解压完成。")
    except Exception as e:
        print(f"[ERROR] 解压失败: {e}")

def extract_zip(file_path, extract_path):
    """
    解压 .zip 文件到指定目录。
    """
    if not os.path.exists(file_path):
        return

    print(f"[INFO] 正在解压 {file_path} 到 {extract_path} ...")
    try:
        with zipfile.ZipFile(file_path, 'r') as zip_ref:
            zip_ref.extractall(extract_path)
        print("[INFO] 解压完成。")
    except Exception as e:
        print(f"[ERROR] 解压失败: {e}")

def download_cadets():
    print("\n=== 处理 Cadets 数据集 ===")
    dataset_name = "cadets"
    dataset_dir = os.path.join(BASE_DIR, dataset_name)
    
    # 文件名根据 notebook 中的 tar 命令推断
    files = [
        ("https://drive.google.com/file/d/1AcWrYiBmgAqp7DizclKJYYJJBQbnDMfb/view?usp=drive_link", "ta1-cadets-e3-official.json.tar.gz"),
        ("https://drive.google.com/file/d/1EycO23tEvZVnN3VxOHZ7gdbSCwqEZTI1/view?usp=drive_link", "ta1-cadets-e3-official-2.json.tar.gz")
    ]
    
    for url, filename in files:
        file_path = os.path.join(dataset_dir, filename)
        download_from_drive(url, file_path)
        extract_tar_gz(file_path, dataset_dir)

def download_fivedirections():
    print("\n=== 处理 Fivedirections 数据集 ===")
    dataset_name = "fivedirections"
    dataset_dir = os.path.join(BASE_DIR, dataset_name)
    
    files = [
        ("https://drive.google.com/file/d/1BeP80zUUmm4eZl0UuU43PsKNkl_xgskj/view?usp=drive_link", "ta1-fivedirections-e3-official-2.json.tar.gz")
    ]
    
    for url, filename in files:
        file_path = os.path.join(dataset_dir, filename)
        download_from_drive(url, file_path)
        extract_tar_gz(file_path, dataset_dir)

def download_optc():
    print("\n=== 处理 OpTC 数据集 ===")
    dataset_name = "optc"
    dataset_dir = os.path.join(BASE_DIR, dataset_name)
    
    if not os.path.exists(dataset_dir):
        os.makedirs(dataset_dir)
        
    # OpTC 的文件较多，且 gdown 会自动识别文件名。
    # 这里我们切换工作目录到 dataset_dir 让 gdown 直接下载到该目录。
    urls = [
        "https://drive.google.com/file/d/1HFSyvmgH0jvdnnnTdKfWRjZYOrLWoIkv/view?usp=drive_link",
        "https://drive.google.com/file/d/1pJLxJsDV8sngiedbfVajMetczIgM3PQd/view?usp=drive_link",
        "https://drive.google.com/file/d/1fRQqc68r8-z5BL7H_eAKIDOeHp7okDuM/view?usp=drive_link",
        "https://drive.google.com/file/d/1VfyGr8wfSe8LBIHBWuYBlU8c2CyEgO5C/view?usp=drive_link",
        "https://drive.google.com/file/d/10N9ZPolq_L8HivBqzf_jFKbwjSxddsZp/view?usp=drive_link",
        "https://drive.google.com/file/d/1xIr8gw-4zc8ESjUpYtrFsbOwhPGUSd15/view?usp=drive_link",
        "https://drive.google.com/file/d/1PvlCp2oQaxEBEFGSQWfcFVj19zLOe7yH/view?usp=drive_link"
    ]
    
    cwd = os.getcwd()
    try:
        os.chdir(dataset_dir)
        for url in urls:
            print(f"[INFO] 正在处理 URL: {url}")
            # gdown 会自动检查文件是否已存在（如果文件名匹配），或者我们可以依赖它的缓存机制
            gdown.download(url, quiet=False, use_cookies=False, fuzzy=True, resume=True)
    except Exception as e:
        print(f"[ERROR] OpTC 下载出错: {e}")
    finally:
        os.chdir(cwd)

def download_theia():
    print("\n=== 处理 Theia 数据集 ===")
    dataset_name = "theia"
    dataset_dir = os.path.join(BASE_DIR, dataset_name)
    
    files = [
        ("https://drive.google.com/file/d/10cecNtR3VsHfV0N-gNEeoVeB89kCnse5/view?usp=drive_link", "ta1-theia-e3-official-1r.json.tar.gz"),
        ("https://drive.google.com/file/d/1Kadc6CUTb4opVSDE4x6RFFnEy0P1cRp0/view?usp=drive_link", "ta1-theia-e3-official-6r.json.tar.gz")
    ]
    
    for url, filename in files:
        file_path = os.path.join(dataset_dir, filename)
        download_from_drive(url, file_path)
        extract_tar_gz(file_path, dataset_dir)

def download_trace():
    print("\n=== 处理 Trace 数据集 ===")
    dataset_name = "trace"
    dataset_dir = os.path.join(BASE_DIR, dataset_name)
    
    files = [
        ("https://drive.google.com/file/d/1GG1aUnPjjzzdbxznVTN8X6oVfA-K4oIV/view?usp=drive_link", "ta1-trace-e3-official-1.json.tar.gz")
    ]
    
    for url, filename in files:
        file_path = os.path.join(dataset_dir, filename)
        download_from_drive(url, file_path)
        extract_tar_gz(file_path, dataset_dir)

def download_streamspot():
    print("\n=== 处理 Streamspot 数据集 ===")
    dataset_name = "streamspot"
    dataset_dir = os.path.join(BASE_DIR, dataset_name)
    
    # Streamspot 托管在 GitHub，使用 git clone
    if os.path.exists(dataset_dir):
        print(f"[INFO] {dataset_dir} 目录已存在，跳过 Clone。")
    else:
        print(f"[INFO] 正在 Clone Streamspot 仓库...")
        try:
            subprocess.run(["git", "clone", "https://github.com/sbustreamspot/sbustreamspot-data", dataset_dir], check=True)
        except Exception as e:
            print(f"[ERROR] git clone 失败: {e}")
            return

    # 检查并解压仓库内的 tar.gz 文件
    tar_path = os.path.join(dataset_dir, "all.tar.gz")
    if os.path.exists(tar_path):
        extract_tar_gz(tar_path, dataset_dir)

def download_unicorn():
    print("\n=== 处理 Unicorn 数据集 ===")
    dataset_name = "unicorn"
    dataset_dir = os.path.join(BASE_DIR, dataset_name)
    
    files = [
        ("https://drive.google.com/file/d/1dmezgT9zQ-8ydHrXvJmxwJN-LjPtumbU/view", "unicorn.zip")
    ]
    
    for url, filename in files:
        file_path = os.path.join(dataset_dir, filename)
        download_from_drive(url, file_path)
        extract_zip(file_path, dataset_dir)

# 定义基础数据集目录
BASE_DIR = "datasets"
if not os.path.exists(BASE_DIR):
    os.makedirs(BASE_DIR)

print(f"开始下载所有数据集到 {BASE_DIR}/ 目录...")
download_cadets()
download_fivedirections()
download_optc()
download_theia()
download_trace()
download_streamspot()
download_unicorn()
```
